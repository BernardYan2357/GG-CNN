## 入门示例

- 按住Ctrl+左键移动鼠标旋转视角
- 滑动滚轮缩放视角
- 左键拖动物体

````python
import pybullet as p
import pybullet_data # 导入PyBullet数据包，包含一些预定义的URDF文件和资源
import time # 导入time模块，用于控制仿真步长


while p.isConnected():
    p.disconnect() # 这一步是防止之前的连接没有正确断开，导致报错 Only one local in-process GUI/GUI_SERVER connection allowed.
physicsClient = p.connect(p.GUI) # 连接到PyBullet仿真环境，使用GUI模式，也可以设置为p.DIRECT
p.setAdditionalSearchPath(pybullet_data.getDataPath()) # 设置额外的搜索路径，当指定的路径中没有找到资源时，PyBullet会在这个路径中查找
# ybullet_data.getDataPath() = ~\anaconda3\envs\ggcnn_env\lib\site-packages\pybullet_data，在该路径下包含了PyBullet预定义的资源

p.setGravity(0, 0, -9.81) # 设置重力加速度
planeId = p.loadURDF("plane.urdf") # 加载平面
# 加载panda机械臂
cubestartPos = [0,0,0] # 定义机器人模型的初始位置，坐标为(x, y, z)
cubeStartorientation = p.getQuaternionFromEuler([0,0,0]) # 定义机器人模型的初始姿态，使用欧拉角表示，并将其转换为四元数格式
boxId = p.loadURDF("franka_panda/panda.urdf", cubestartPos, cubeStartorientation) # 加载URDF文件
for i in range (10000): # 运行仿真10000步
    p.stepSimulation() # 进行一步仿真，更新物理状态
    time.sleep(1./240.) # 暂停一段时间，控制仿真步长为1/240秒
    cubePos, cubeOrn = p.getBasePositionAndOrientation(boxId) # 获取机器人模型的当前位姿，包括位置和姿态
    # print(cubePos,cubeOrn)
p.disconnect() # 断开与PyBullet仿真环境的连接
````
